3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**.

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
#ваш код здесь

def get_experience(arg):

    month_keys = ["месяц", "месяца", "месяцев"]
    year_key = ["год", "года", "лет"]

    year = 0
    month = 0

    if pd.isnull(arg) or arg == "Не указано":
        return np.nan

    else:
        arg_split = arg.split("  ")[0].split(" ")[2:]

        for i in range(len(arg_split)):

            if arg_split[i] in month_keys:
                month = arg_split[i - 1]

            if arg_split[i] in year_key:
                year = arg_split[i - 1]

        return int(year) * 12 + int(month)


database["Опыт работы (месяц)"] = database["Опыт работы"].apply(get_experience)

del database["Опыт работы"]

```

После ответа на контрольные вопросы удалите исходный столбец заработной платы "ЗП" и все промежуточные столбцы, если вы их создавали.

Итак, давайте обсудим возможный алгоритм преобразования:
1. Перевести признак "Обновление резюме" из таблицы с резюме в формат datetime и достать из него дату. В тот же формат привести признак "date" из таблицы с валютами.
2. Выделить из столбца "ЗП" сумму желаемой заработной платы и наименование валюты, в которой она исчисляется. Наименование валюты перевести в стандарт ISO согласно с таблицей выше.
3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
#ваш код здесь

exchange_rates = pd.read_csv("sample_data/ExchangeRates.csv")

currency_dict = {
    "грн": "UAH",
    "USD": "USD",
    "EUR": "EUR",
    "белруб": "BYN",
    "KGS": "KGS",
    "сум": "UZS",
    "AZN": "AZH",
    "KZT": "KZT",
}

# 1. Обновить признак "Обновление резюме" из таблицы с резюме, переведя его в формат datetime и выделив из него дату.
# В тот же формат привести признак date из таблицы с валютами.
exchange_rates["date"] = pd.to_datetime(exchange_rates["date"], dayfirst=True).dt.date
database["Обновление резюме"] = pd.to_datetime(
    database["Обновление резюме"], dayfirst=True
).dt.date

# 2. Выделить из столбца «ЗП» сумму желаемой заработной платы и наименование валюты, в которой она исчисляется
database[["ЗП (сум)", "ЗП (валюта)"]] = database["ЗП"].str.split(" ", expand=True)
database["ЗП (валюта)"] = (
    database["ЗП (валюта)"].str.replace(".", "", regex=False).str.replace(" ", "")
)
database["ЗП (валюта)"] = database["ЗП (валюта)"].replace(currency_dict)

# 3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты
# Значение close для рубля заполнить единицей 1 (курс рубля самого к себе).

database = pd.merge(
    database,
    exchange_rates,
    how="left",
    right_on=["date", "currency"],
    left_on=["Обновление резюме", "ЗП (валюта)"],
)

database.loc[database["ЗП (валюта)"] == "руб", "close"] = 1
database.loc[database["ЗП (валюта)"] == "руб", "proportion"] = 1

database["ЗП (руб)"] = database["ЗП (сум)"].astype(float) * database["close"] / database["proportion"]

# Удалите исходный столбец заработной платы «ЗП» и все промежуточные столбцы, если вы их создавали.
del database["ЗП (валюта)"], database["ЗП (сум)"], database["ЗП"],
```

3. Итак, у нас есть пропуски в 3ех столбцах: **"Опыт работы (месяц)"**, **"Последнее/нынешнее место работы"**, **"Последняя/нынешняя должность"**. Поступим следующим образом: удалите строки, где есть пропуск в столбцах с местом работы и должностью. Пропуски в столбце с опытом работы заполните **медианным** значением.

```python
import pandas as pd
import re

# Загрузка данных
database = pd.read_csv("/content/sample_data/dst-3.0_16_1_hh_database.csv", sep=";")

# Удаляем строки с пропущенными данными в указанных столбцах
clear_db = database.dropna(
    subset=["Последнее/нынешнее место работы", "Последняя/нынешняя должность"]
).copy()

# Функция для извлечения опыта работы в месяцах
def extract_experience(text):
    if pd.isna(text):
        return None
    match_years = re.search(r'(\d+)\s+лет', text)
    match_months = re.search(r'(\d+)\s+месяц', text)
    years = int(match_years.group(1)) if match_years else 0
    months = int(match_months.group(1)) if match_months else 0
    return years * 12 + months

# Преобразуем столбец "Опыт работы" в числовой формат
clear_db.loc[:, "Опыт работы (месяц)"] = clear_db["Опыт работы"].apply(extract_experience)

# Заполняем пропущенные значения медианой
median_experience = clear_db["Опыт работы (месяц)"].median()
clear_db.loc[:, "Опыт работы (месяц)"] = clear_db["Опыт работы (месяц)"].fillna(median_experience)

# Выводим средний опыт работы в месяцах
print(f"Средний опыт работы: {int(clear_db['Опыт работы (месяц)'].mean())} месяцев")
```

```python
data_cleaned = database.dropna(subset=['Последнее/нынешнее место работы', 'Последняя/нынешняя должность']).copy()
```

```python
print(data_cleaned.columns)
```

```python
import re

# Функция для извлечения опыта работы в месяцах
def extract_experience(text):
    if pd.isna(text):
        return None
    # Извлекаем количество лет и месяцев
    match_years = re.search(r'(\d+)\s+лет', text)
    match_months = re.search(r'(\d+)\s+месяц', text)
    years = int(match_years.group(1)) if match_years else 0
    months = int(match_months.group(1)) if match_months else 0
    return years * 12 + months

# Создаем новый столбец "Опыт работы (месяц)"
if "Опыт работы" in data_cleaned.columns:
    data_cleaned["Опыт работы (месяц)"] = data_cleaned["Опыт работы"].apply(extract_experience)
else:
    print("Столбец 'Опыт работы' отсутствует в данных.")

# Заполняем пропущенные значения медианой
if "Опыт работы (месяц)" in data_cleaned.columns:
    median_experience = data_cleaned['Опыт работы (месяц)'].median()
    data_cleaned['Опыт работы (месяц)'] = data_cleaned['Опыт работы (месяц)'].fillna(median_experience)
    print("Пропуски в 'Опыт работы (месяц)' заполнены медианой.")
else:
    print("Столбец 'Опыт работы (месяц)' отсутствует. Проверьте данные.")
```

```python
# Проверяем наличие столбца "ЗП"
if "ЗП" in data_cleaned.columns:
    # Преобразуем значения в числовой формат (на случай, если это еще не сделано)
    data_cleaned["ЗП"] = pd.to_numeric(data_cleaned["ЗП"], errors="coerce")

    # Выводим количество записей с выбросами
    num_outliers = data_cleaned[
        (data_cleaned["ЗП"] > 10**6) | (data_cleaned["ЗП"] < 1000)
    ].shape[0]
    print(f"Количество резюме с выбросами по зарплате: {num_outliers}")

    # Удаляем выбросы
    data_cleaned = data_cleaned[
        ~((data_cleaned["ЗП"] > 10**6) | (data_cleaned["ЗП"] < 1000))
    ].copy()
    print(f"Осталось записей после очистки: {data_cleaned.shape[0]}")
else:
    print("Столбец 'ЗП' отсутствует в данных.")
```

```python
# ваш код здесь

import pandas as pd
import re

# Функция для извлечения возраста
def extract_age(text):
    if pd.isna(text):
        return None
    match = re.search(r'(\d+)', text)
    return int(match.group(1)) if match else None

# Убедимся, что возраст соискателя рассчитан
if "Пол, возраст" in data_cleaned.columns:
    data_cleaned["Возраст"] = data_cleaned["Пол, возраст"].apply(extract_age)
else:
    print("Столбец 'Пол, возраст' отсутствует в данных.")

# Проверяем, что данные о возрасте и опыте работы готовы
if "Возраст" in data_cleaned.columns and "Опыт работы (месяц)" in data_cleaned.columns:
    # Преобразуем опыт работы в года
    data_cleaned["Опыт работы (годы)"] = data_cleaned["Опыт работы (месяц)"] / 12

    # Находим резюме, где опыт работы превышает возраст
    invalid_entries = data_cleaned[data_cleaned["Опыт работы (годы)"] > data_cleaned["Возраст"]]
    print(f"Найдено резюме с некорректными данными: {invalid_entries.shape[0]}")

    # Удаляем такие записи
    data_cleaned = data_cleaned[data_cleaned["Опыт работы (годы)"] <= data_cleaned["Возраст"]].copy()
    print(f"Осталось записей после удаления некорректных данных: {data_cleaned.shape[0]}")
else:
    print("Не удалось выполнить расчет: проверьте наличие столбцов 'Возраст' и 'Опыт работы (месяц)'.")
```

Попробуйте построить распределение признака в **логарифмическом масштабе**. Добавьте к графику линии, отображающие **среднее и границы интервала метода трех сигм**. В какую сторону асимметрично логарифмическое распределение? Напишите об этом в комментарии к графику.
Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. Давайте сделаем послабление на **1 сигму** (возьмите 4 сигмы) в **правую сторону**.

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re

# Загружаем данные
file_path = '/content/sample_data/dst-3.0_16_1_hh_database.csv'
data = pd.read_csv(file_path, sep=';')

# Извлекаем возраст
def extract_age(text):
    if pd.isna(text):
        return None
    match = re.search(r'(\d+)', text)
    return int(match.group(1)) if match else None

data['Возраст'] = data['Пол, возраст'].apply(extract_age)

# Преобразуем возраст в логарифмический масштаб
data['log_age'] = np.log(data['Возраст'])

# Рассчитываем статистики
log_age_mean = data['log_age'].mean()
log_age_std = data['log_age'].std()

# Границы метода трех сигм (с ослаблением в правую сторону до 4 сигм)
lower_bound = log_age_mean - 3 * log_age_std
upper_bound = log_age_mean + 4 * log_age_std

# Строим гистограмму логарифмического распределения возраста
plt.figure(figsize=(10, 6))
histplot = sns.histplot(data['log_age'], bins=30, kde=True, color='blue', edgecolor='black', alpha=0.7)
histplot.axvline(log_age_mean, color='k', lw=2, label='Среднее')
histplot.axvline(lower_bound, color='red', lw=2, linestyle='--', label='Нижняя граница (3σ)')
histplot.axvline(upper_bound, color='green', lw=2, linestyle='--', label='Верхняя граница (4σ)')

plt.title('Распределение возраста в логарифмическом масштабе с границами метода 3 сигм', fontsize=14)
plt.xlabel('Логарифм возраста', fontsize=12)
plt.ylabel('Частота', fontsize=12)
plt.legend(fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Найдем выбросы
outliers = data[(data['log_age'] < lower_bound) | (data['log_age'] > upper_bound)]

# Удалим выбросы
data_cleaned = data[~((data['log_age'] < lower_bound) | (data['log_age'] > upper_bound))].copy()

# Выводим таблицу выбросов
print("Выбросы по возрасту:")
print(outliers[['Пол, возраст', 'Возраст', 'log_age']])

# Оценим возраст выбросов
outlier_ages = np.exp(outliers['log_age'])
print(f"Возраст выбросов (в годах): {outlier_ages.values}")
```

```python
# Построим график для очищенных данных (cleaned_data)
plt.figure(figsize=(10, 6))
histplot_cleaned = sns.histplot(data_cleaned['log_age'], bins=30, kde=True, color='blue', edgecolor='black', alpha=0.7)
histplot_cleaned.axvline(log_age_mean, color='k', lw=2, label='Среднее')
histplot_cleaned.axvline(lower_bound, color='red', lw=2, linestyle='--', label='Нижняя граница (3σ)')
histplot_cleaned.axvline(upper_bound, color='green', lw=2, linestyle='--', label='Верхняя граница (4σ)')

plt.title('Распределение возраста в логарифмическом масштабе (очищенные данные)', fontsize=14)
plt.xlabel('Логарифм возраста', fontsize=12)
plt.ylabel('Частота', fontsize=12)
plt.legend(fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
